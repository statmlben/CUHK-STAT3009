### **Importing Data based on pandas**

#### Importing Data from an CSV File

**Task:** Import a CSV file into a pandas DataFrame.

**Instructions:**
1. Download a sample CSV file (e.g., from [Kaggle Datasets](https://www.kaggle.com/datasets)).
2. Use pandas to read the CSV file and display the first 5 rows.

**Sample Code:**
```python
import pandas as pd

# Replace 'your_file.csv' with the path to your CSV file
df = pd.read_csv('your_file.csv')
print(df.head())
```

#### Importing Data from an Excel File

**Task:** Import data from an Excel file into a pandas DataFrame.

**Instructions:**
1. Download an Excel file (e.g., from [Kaggle Datasets](https://www.kaggle.com/datasets)).
2. Use pandas to read a specific sheet and display the first 5 rows.

**Sample Code:**
```python
import pandas as pd

# Replace 'your_file.xlsx' with the path to your Excel file
df = pd.read_excel('your_file.xlsx', sheet_name='Sheet1')
print(df.head())
```

#### Importing Data from a URL (CSV)

**Task:** Import data directly from a URL that points to a CSV file.

**Instructions:**
1. Find a public CSV file URL (for example, from [data.gov](https://www.data.gov/)).
2. Use pandas to read the CSV file from the URL and display the first 5 rows.

**Sample Code:**
```python
import pandas as pd

# Replace 'http://example.com/your_file.csv' with a valid CSV URL
url = 'http://example.com/your_file.csv'
df = pd.read_csv(url)
print(df.head())
```

#### Importing Data with Custom Delimiter

**Task:** Import a CSV file with a custom delimiter (e.g., semicolon).

**Instructions:**
1. Use a CSV file that uses a semicolon as a delimiter.
2. Read the file using the appropriate parameter.

**Sample Code:**
```python
import pandas as pd

# Replace 'your_file.csv' with the path to your CSV file
df = pd.read_csv('your_file.csv', delimiter=';')

```
